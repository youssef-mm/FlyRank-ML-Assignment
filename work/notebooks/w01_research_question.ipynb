## 2. The question: decision, action, cost of a wrong call

Research Question: Which existing content pages with declining trends can be automatically flagged for refreshing to recover lost search visibility?

Decision: Deciding whether a specific underperforming page is worth a content update or editorial rewrite.

## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import pandas as pd

url = "https://raw.githubusercontent.com/youssef-mm/FlyRank-ML-Assignment/main/data/raw/content_refresh_anonymized.csv"
df = pd.read_csv(url)

total_pages = df.shape[0]
# حساب نسبة الصفحات التي لديها اتجاه هبوطي بناءً على عمود trend_direction
declining_rate = round((df["trend_direction"] == "declining").mean() * 100, 2)
avg_impressions = round(df["impressions_90d"].mean(), 2)

print(f"Total pages in dataset: {total_pages}")
print(f"Percentage of declining pages: {declining_rate}%")
print(f"Average 90-day impressions per page: {avg_impressions}")
```
