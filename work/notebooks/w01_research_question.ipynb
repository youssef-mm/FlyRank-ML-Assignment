**Why this lane:**
In search engine optimization and digital publishing, reviving existing decaying high-exposure content delivers a substantially higher return on investment than authoring new content from scratch. ## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

### Research Question
**Which existing content pages with declining organic search visibility represent the highest-priority candidates for editorial refresh to recover lost traffic?**

### Core Operational Framework
- **Unit of Analysis:** One row = one unique content page (`content_id`) for a specific client (`client_id`) evaluated over a trailing 90-day observation window.
- **Decision:** Deciding whether a specific underperforming page justifies editorial resources for an update, rewrite, and keyword re-optimization, or should be left alone / monitored.
- **Who Acts on It & Action Taken:** Content editors and SEO managers. They review the prioritized queue and execute targeted on-page refreshes: updating outdated statistics, addressing intent shifts, expanding thin sections, restructuring headers, and refreshing metadata.
- **Output:** A ranked review queue of content pages prioritized by refresh urgency, accompanied by interpretable reason codes (e.g., `stale_visible_decay_risk`) and model confidence tiers.

### Cost of a Wrong Call
- **False Positive (recommending a healthy, stable, or dead page):** Wastes scarce editorial hours rewriting content that does not need intervention, incurs an opportunity cost by delaying updates on truly decaying high-volume pages, and risks breaking search rankings on pages that were already performing well.
- **False Negative (missing a truly declining page):** Leads to compounding organic search traffic loss, erosion of keyword rankings to competitors, and delayed intervention.

### Why Data and ML Help at All
A simple fixed rule (e.g., "refresh content if age > 180 days") fails because evergreen articles often sustain high rankings for years without updates, while newer pages in competitive topics may experience sharp decay in under 60 days. Machine learning excels here because it simultaneously models non-linear interactions across historical impression volume (`impressions_90d`), ranking position decay (`avg_position`), CTR deficits, and update age (`days_since_last_update`).

## 3. Quick look at the data (2-3 real numbers)

*Load the starter CSV below and show 2-3 real numbers that make your lane look worth the next 7 weeks.*

```python
import os
import pandas as pd

# Load dataset (local path with Colab raw URL fallback)
data_path = "data/raw/content_refresh_anonymized.csv"
if not os.path.exists(data_path):
    data_path = "../../data/raw/content_refresh_anonymized.csv"
if not os.path.exists(data_path):
    data_path = "https://raw.githubusercontent.com/youssef-mm/FlyRank-ML-Assignment/main/data/raw/content_refresh_anonymized.csv"

df = pd.read_csv(data_path)

total_pages = len(df)
total_clients = df["client_id"].nunique()
declining_count = (df["trend_direction"] == "down").sum()
declining_rate = round((df["trend_direction"] == "down").mean() * 100, 2)
avg_impressions = round(df["impressions_90d"].mean(), 2)
median_impressions = round(df["impressions_90d"].median(), 2)

# High-leverage candidate slice: stale visible content (>= 90 days since update, >= 500 impressions)
stale_visible = df[(df["days_since_last_update"] >= 90) & (df["impressions_90d"] >= 500)]
stale_visible_count = len(stale_visible)
stale_visible_declining_rate = round((stale_visible["trend_direction"] == "down").mean() * 100, 2)

print(f"1. Total content inventory: {total_pages:,} pages across {total_clients} clients.")
print(f"2. Overall decline rate (trend_direction == 'down'): {declining_rate}% ({declining_count:,} pages).")
print(f"3. 90-day search impressions: Mean = {avg_impressions:,.2f} | Median = {median_impressions:,.2f} (heavy-tailed distribution).")
print(f"4. Stale visible opportunity pool (update >= 90d & imp >= 500): {stale_visible_count:,} pages ({stale_visible_count/total_pages*100:.1f}% of inventory).")
print(f"5. Decline rate in stale visible pool: {stale_visible_declining_rate}% (+{stale_visible_declining_rate - declining_rate:.2f}% vs baseline).")
```

### What I can claim:
